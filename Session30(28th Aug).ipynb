# Missing Value Treatment For Numerical Data

```python
clean_data.isnull().sum()
```

```python
clean_data['Age']
```

```python
import numpy as np
```

```python
clean_data["Age"]=clean_data["Age"].fillna(np.mean(pd.to_numeric(clean_data["Age"])))
```

```python
clean_data["Age"]
```

```python
clean_data["Exp"]
```

```python
clean_data["Exp"]=clean_data["Exp"].fillna(np.mean(pd.to_numeric(clean_data["Exp"])))
```

```python
clean_data["Exp"]
```

```python
clean_data
```

```python
clean_data["Location"].isnull().sum()
```

```python
clean_data["Location"]=clean_data["Location"].fillna(clean_data["Location"].mode()[0])
```

#Select the column from the data frame location column has text values so it implements mode strategy to handle missing values if we have most frequent values more than one we select the first value that's why we use mode()[0] otherwise mode()
