```python
i=1 #initial value
while i<=5: # condition
    print('data science')
    i=i+1 # increment value
```

```python
i=5 #initial value
while i>=1: # condition
    print('data science :',i)
    i=i-1 # Decrement value
```

```python
i=1
while i<=5:
    print('data science') #when we mention end then new line will not print
    j=1
    while j<=4:
        print('technology')
        j=j+1
        
    i=i+1
    print()
```

```python
i=1
while i<=5:
    print('data science',end=' ') #when we mention end then new line will not print
    j=1
    while j<=4:
        print('technology',end=' ')
        j=j+1
        
    i=i+1
    print()
```
