```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load sample dataset
tips=sns.load_dataset('tips')

# set visual theme
sns.set_theme(style='darkgrid')
```

```python
plt.figure(figsize=(8,6)) # 8 is width and 6 is height
```

```python
# 1.Scatter Plot: Total Bill vs Tip
sns.scatterplot(data=tips,x="total_bill",y="tip",hue="time",size="size",palette="deep")
plt.title("Scatter plot:Total_bill vs Tip by Time and Size")
plt.show()
```

```python
# 2. Line Plot: Total Bill by Party Size,split by sex
sns.lineplot(data=tips,x="size",y="total_bill",hue="sex",ci=None,marker="o")
plt.title("Line plot:Total_bill by party Size and Sex")
plt.show()
```

```python
# Line Plot
sns.lineplot(data=tips,x="size",y="total_bill",hue="sex",ci=95,marker="o")
plt.title("Line plot:Total_bill by party Size and Sex")
plt.show()
```

```python
# Bar Plot: Average Total_bill by Day and split by Sex
sns.barplot(data=tips,x="day",y="total_bill",hue="sex",palette="muted") #muted-less colorful
plt.title("Bar plot:Average Total_bill by Day and Sex")
plt.show()
```

```python
# Box plot: Total Bill by day,split by time
sns.boxplot(data=tips,x="day",y="tip",hue="smoker",palette="Set2")
plt.title("Box plot:Tip by Day and Smoker Status")
plt.show()
```

```python
#Violin Graph/Plot: Total Bill by day,split by time
sns.violinplot(data=tips,x="day",y="total_bill",hue="time",split=True,palette="pastel")
plt.title("Violin plot:Total_bill by Day and Time")
plt.show()
```

```python
#Violin Graph/Plot
sns.violinplot(data=tips,x="day",y="total_bill",hue="time",split=False,palette="pastel")
plt.show()
```

```python
# Count Plot
sns.countplot(data=tips,x="day",hue="smoker",palette="dark")
plt.title("Count plot:Orders by Day and Smoker Status")
plt.show()
```

```python
#Regression plot
sns.regplot(data=tips,x="total_bill",y="tip",scatter_kws={"s":50},line_kws={"color":"red"})
plt.title("Regression plot:Total_bill vs Tip")
plt.show()
```

```python
# Histogram (Distribution Plot):Total Bill with KDE
sns.histplot(data=tips,x="total_bill",kde=True,bins=20,color="blue")
plt.title("Histogram:Distribution of total Bill")
plt.show()
```

```python
# Pair Plot: Relationship Between numerical variable
sns.pairplot(tips, hue="sex", vars=["total_bill", "tip", "size"], palette="husl")

# Correct: suptitle instead of subtitle
plt.suptitle("Pair Plot: Numerical variable by Sex", y=1.02)

plt.show()
```

```python
# CatPlot(Point Plot): Tip by Day and Sex
sns.catplot(data=tips,x="day",y="tip",hue="sex",kind="point",palette="bright")
plt.title("Catplot(Point):Tip by Day and Sex")
plt.show()
```

```python
#Joint Plot:Total Bill vs Tip with marginal Distributions
sns.jointplot(data=tips,x="total_bill",y="tip",kind="scatter",hue="smoker",palette="coolwarm")
plt.suptitle("Joint Plot:Total Bill vs Tip by smoker",y=1.02)
plt.show()
```

```python
# FacetGrid:Total Bill by day, Facted by Time and Smoker
g=sns.FacetGrid(tips,col="time",row="smoker",margin_titles=True)
g.map(sns.histplot,"total_bill",bins=15)
g.fig.suptitle("FacetGrid:Total Bill by Time and Smoker",y=1.02)
plt.show()
```

```python
# Strip Plot: Tips by Day, colored by sex
plt.figure(figsize=(8,6))
sns.stripplot(data=tips, x="day", y="tip", hue="sex", palette="Set1", jitter=True)
plt.title("Strip Plot: Tip by Day and Sex")
plt.show()
```

```python
# KDE Plot:Total Bill density by Sex
plt.figure(figsize=(8,6))
sns.kdeplot(data=tips, x="total_bill", hue="sex",fill=True, palette="tab10")
plt.title("KDE Plot:Total Bill Density by Sex")
plt.show()
```

#Explanations of Plot
-> Scatter Plot-: Shows RelationShip Between total_bill and tip,with time and size as additional dimension
-> LinePlot-: Trends of total_bill by day ,split by sex
-> BarPlot-: Average total_bill by day,split by sex
-> BoxPlot-: Distribution of tip by day,split by smoker
-> Violin Plot-: Distribution of total bill by day,split by time
-> Count Plot-: Counts of Orders by day,split by smoker
-> Regression Plot-: Linear fit between total bill and tip
-> Histogram-: Distribution of total_bill with a KDE curve
-> Pair Plot-: PairWise relationship between total bill,tip,and size,colourby sex
