```python
def add_sub(x,y):
    c=x+y
    d=x-y
    return c,d  
result,result1=add_sub(4,5)
print(result)
print(result1) 
print(type(result))  
print(type(result1))
```

```python
def add_sub_mul(x, y):
    c = x + y
    d = x - y
    e = x * y
    return c, d, e

result, result1, result2 = add_sub_mul(4,5)

print(result)   # sum
print(result1)  # difference
print(result2)  # product

print(type(result))  
print(type(result1))
print(type(result2))
```

# Function Arguments
1. Formal Argument
2. Actual Argument

```python
def update():
    x=8
    print(x)
update()
```

```python
def add(a,b): # a and b are called as formal arguments
    c=a+b
    print(c)
add(5,6) # 5 & 6 are called as actual arguments
```

# KeyWord Argument

```python
def person(name,age):
    print(name)
    print(age+1)
person('nit',22)
```

```python
def person(name,age):
    print(name)
    print(age)
person(22, 'nit')
```

```python
def person(name,age):
    print(name)
    print(age+1)
person(22, 'nit')
```

```python
def person(name,age): # this function is called as keyword arguments
    print(name)
    print(age+1)
person(age=22, name= 'nit')
```

```python
def person(name,age): # this function is called as keyword arguments
    print(name)
    print(age+1)
person(age=22, name= 'nit',phone=928461)
```

```python
def person(name,age,phone): # this function is called as keyword arguments
    print(name)
    print(age+1)
    print(phone)
person(age=22, name= 'nit',phone=928461)
```
